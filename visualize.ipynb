# ShowerLM showers in 3D

Reads the `sample_*.npz` files written by `generate.py` and draws them in 3D (light accumulated along the line of sight, box in metres, one colour scale for all panels).

```python
import glob
import matplotlib.pyplot as plt
from showerlm.viz import gallery

files = sorted(glob.glob("showers/sample_*.npz"))[:1]
print(len(files), "showers")
```

```python
fig = gallery(files, cols=4)
plt.show()
```

Save the figure:

```python
fig.savefig("showers_3d.png", dpi=150, facecolor=fig.get_facecolor())
```
